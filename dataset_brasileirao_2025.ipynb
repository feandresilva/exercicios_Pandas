```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
pd.set_option('display.max_columns', None)

df_brasileirao = pd.read_excel("https://github.com/schneider8357/projeto_brasileirao/raw/refs/heads/main/dados_brasileirao_serie_a_2025.xlsx")
df_brasileirao
```

Monte a tabela final do campeonato conforme a seguinte regra:

-Primeiro ordenado pelo maior número de pontos (3 pontos cada vitória, 1 ponto cada empate)

-Depois ordenado pelo Maior saldo de gols

-E por último, ordenado pelo maior número de gols pró

Faça um programa que receba como entrada o nome de um time e mostre em gráficos a evolução de um time pelo campeonato.

- Evolução de gols feitos
- Evolução de gols tomados
- Evolução de saldo de gols

- Percentual de vitórias vs empates vs derrotas

Ordene pelo número da rodada.

```python
#Qual time teve o maior número de gols?
df_gols_time1 = df_brasileirao[["nome_time1", "gols_time1"]]
df_gols_time2 = df_brasileirao[["nome_time2", "gols_time2"]]
df_gols_time1 = df_gols_time1.rename(columns={"nome_time1": "nome_time", "gols_time1": "gols_time"})
df_gols_time2 = df_gols_time2.rename(columns={"nome_time2": "nome_time", "gols_time2": "gols_time"})
df_brasileirao_unificado = pd.concat([df_gols_time1, df_gols_time2])
df_brasileiro_unificado = df_brasileirao_unificado.groupby("nome_time")["gols_time"].sum().reset_index()
df_brasileiro_unificado = df_brasileiro_unificado.rename(columns={"gols_time": "gols_pro"})
df_brasileiro_unificado
```

```python
#Qual time teve o maior saldo de gols?
df_saldo_time1 = df_brasileirao[["nome_time1", "gols_time1", "gols_time2"]]
df_saldo_time1["saldo"] = df_saldo_time1["gols_time1"] - df_saldo_time1["gols_time2"]
df_saldo_time1 = df_saldo_time1.rename(columns={"nome_time1": "nome_time"})
df_saldo_filtrado1 = df_saldo_time1[["nome_time", "saldo"]]
df_saldo_time2 = df_brasileirao[["nome_time2", "gols_time2", "gols_time1"]]
df_saldo_time2["saldo"] = df_saldo_time2["gols_time2"] - df_saldo_time2["gols_time1"]
df_saldo_time2 = df_saldo_time2.rename(columns={"nome_time2": "nome_time"})
df_saldo_filtrado2 = df_saldo_time2[["nome_time", "saldo"]]
df_saldo_unificado = pd.concat([df_saldo_filtrado1, df_saldo_filtrado2])
df_saldo_unificado = df_saldo_unificado.groupby("nome_time")["saldo"].sum().reset_index()
df_saldo_unificado
```

```python
#Qual time teve o maior número de vitórias?
df_time1 = df_brasileirao[["nome_time1", "gols_time1", "gols_time2"]]
df_vitoria1 = df_time1[df_time1["gols_time1"] > df_time1["gols_time2"]]
df_vitoria1 = df_vitoria1.rename(columns={"nome_time1": "nome_time"})
df_vitoria1_fitlrado = df_vitoria1["nome_time"]
df_time2 = df_brasileirao[["nome_time1", "gols_time1", "gols_time2", "nome_time2"]]
df_vitoria2 = df_time2[df_time2["gols_time2"] > df_time2["gols_time1"]]
df_vitoria2 = df_vitoria2.rename(columns={"nome_time2": "nome_time"})
df_vitoria2_filtrado = df_vitoria2["nome_time"]
df_vitoria_unificado = pd.concat([df_vitoria1_fitlrado, df_vitoria2_filtrado])
df_vitoria_unificado = df_vitoria_unificado.value_counts().reset_index()
df_vitoria_unificado = df_vitoria_unificado.rename(columns={"count": "vitorias"})
df_vitoria_unificado
```

```python
#Quantificando os empates
df_empate1 = df_brasileirao[["nome_time1", "gols_time1", "gols_time2"]]
df_empate1_filtrado = df_empate1[df_empate1["gols_time1"] == df_empate1["gols_time2"]]
df_empate1_filtrado = df_empate1_filtrado.rename(columns={"nome_time1": "nome_time"})
df_empate1_final = df_empate1_filtrado["nome_time"]
df_empate2 = df_brasileirao[["gols_time1", "gols_time2", "nome_time2"]]
df_empate2_filtrado = df_empate2[df_empate2["gols_time1"] == df_empate2["gols_time2"]]
df_empate2_filtrado = df_empate2_filtrado.rename(columns={"nome_time2": "nome_time"})
df_empate2_final = df_empate2_filtrado["nome_time"]
df_empate_unificado = pd.concat([df_empate1_final, df_empate2_final])
df_empate_unificado = df_empate_unificado.value_counts().reset_index()
df_empate_unificado = df_empate_unificado.rename(columns={"count": "empates"})
df_empate_unificado
```

```python
#Monte a tabela final do campeonato conforme a seguinte regra:
#Primeiro ordenado pelo maior número de pontos (3 pontos cada vitória, 1 ponto cada empate)
#Depois ordenado pelo Maior saldo de gols
#E por último, ordenado pelo maior número de gols pró
df_unificado_saldo_vitorias = df_saldo_unificado.set_index("nome_time").join(df_vitoria_unificado.set_index("nome_time"), how="left")
df_unificado_saldo_vitorias = df_unificado_saldo_vitorias.reset_index()
df_tabela_final_intermediaria = df_brasileiro_unificado.set_index("nome_time").join(df_unificado_saldo_vitorias.set_index("nome_time"), how= "left")
df_tabela_final_intermediaria = df_tabela_final_intermediaria.reset_index()
df_tabela_completa = df_empate_unificado.set_index("nome_time").join(df_tabela_final_intermediaria.set_index("nome_time"), how="left")
df_tabela_completa["pontos"] = (df_tabela_completa["vitorias"] * 3) + df_tabela_completa["empates"]
df_tabela_completa["derrotas"] = 38 - (df_tabela_completa["vitorias"] + df_tabela_completa["empates"])
df_tabela_completa["gols_tomados"] = df_tabela_completa["gols_pro"] - df_tabela_completa["saldo"]
df_tabela_completa.sort_values(["pontos", "saldo", "gols_pro"], ascending = False)[["pontos", "vitorias", "empates", "derrotas", "gols_pro", "gols_tomados", "saldo"]]
```

```python
#Faça um programa que receba como entrada o nome de um time e mostre em gráficos a evolução de um time pelo campeonato.
#Exiba em gráficos de linha:
#Evolução de gols feitos
#Evolução de gols tomados
#Evolução de saldo de gols

time_escolhido = input("Escolha um time: ")

df_rodadas1 = df_brasileirao[["rodada", "nome_time1", "gols_time1", "gols_time2", "nome_time2"]]
df_rodadas1["saldo"] = df_rodadas1["gols_time1"] - df_rodadas1["gols_time2"]
df_rodadas1 = df_rodadas1.rename(columns={"nome_time1": "nome_time", "gols_time1": "gols_pro"})
df_rodadas1_filtrado = df_rodadas1[["rodada", "nome_time", "gols_pro", "saldo"]]

df_rodadas2 = df_brasileirao[["rodada", "nome_time1", "gols_time1", "gols_time2", "nome_time2"]]
df_rodadas2["saldo"] = df_rodadas2["gols_time2"] - df_rodadas2["gols_time1"]
df_rodadas2 = df_rodadas2.rename(columns={"nome_time2": "nome_time", "gols_time2": "gols_pro"})
df_rodadas2_filtrado = df_rodadas2[["rodada", "nome_time", "gols_pro", "saldo"]]

df_rodadas_final = pd.concat([df_rodadas2_filtrado, df_rodadas1_filtrado])
df_rodadas_final["gols_tomados"] = df_rodadas_final["gols_pro"] - df_rodadas_final["saldo"]


df_evolucao_gols_feitos = df_rodadas_final.groupby(["rodada", "nome_time"])["gols_pro"].sum()
df_evolucao_gols_feitos = df_evolucao_gols_feitos.reset_index()
df_escolha_usuario_gols_feitos = df_evolucao_gols_feitos[df_evolucao_gols_feitos["nome_time"]== time_escolhido]
fig = px.line(df_escolha_usuario_gols_feitos, x="rodada", y="gols_pro", color="nome_time")
display(fig.show())

df_evolucao_gols_tomados = df_rodadas_final.groupby(["rodada", "nome_time"])["gols_tomados"].sum()
df_evolucao_gols_tomados = df_evolucao_gols_tomados.reset_index()
df_escolha_usuario_gols_tomados = df_evolucao_gols_tomados[df_evolucao_gols_tomados["nome_time"]== time_escolhido]
fig = px.line(df_escolha_usuario_gols_tomados, x="rodada", y="gols_tomados", color="nome_time")
display(fig.show())

df_evolucao_saldo = df_rodadas_final.groupby(["rodada", "nome_time"])["saldo"].sum()
df_evolucao_saldo = df_evolucao_saldo.reset_index()
df_escolha_usuario_saldo = df_evolucao_saldo[df_evolucao_saldo["nome_time"] == time_escolhido]
fig = px.line(df_escolha_usuario_saldo, x="rodada", y="saldo", color="nome_time")
display(fig.show())

df_tabela_completa = df_tabela_completa.reset_index()
df_grafico_pizza = df_tabela_completa[["nome_time", "vitorias", "empates", "derrotas"]]
df_escolha_usuario = df_grafico_pizza[df_grafico_pizza["nome_time"]== time_escolhido]
df_escolha_usuario_percentual = df_escolha_usuario[["vitorias", "empates", "derrotas"]] / 38 *100
df_escolha_usuario_percentual = df_escolha_usuario_percentual.reset_index(drop=True)
df_escolha_usuario_percentual = df_escolha_usuario_percentual.T
df_escolha_usuario_percentual = df_escolha_usuario_percentual.reset_index()
df_escolha_usuario_percentual = df_escolha_usuario_percentual.rename(columns={"index":"Resultado", 0: "Porcentagem"})
fig = px.pie(df_escolha_usuario_percentual, values="Porcentagem", names="Resultado")
display(fig.show())

df_tabela_final_ordenada = df_tabela_completa.sort_values(["pontos", "saldo", "gols_pro"], ascending = False)[["nome_time", "pontos", "vitorias", "empates", "derrotas", "gols_pro", "gols_tomados", "saldo"]]
df_tabela_final_ordenada = df_tabela_final_ordenada.reset_index(drop=True)
df_tabela_final_ordenada["posicao"] = df_tabela_final_ordenada.index + 1
df_escolha_usuario_posicao_final = df_tabela_final_ordenada[df_tabela_final_ordenada["nome_time"]== time_escolhido]
df_escolha_usuario_posicao_final[["nome_time", "posicao"]]

```
