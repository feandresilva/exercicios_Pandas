```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
pd.set_option('display.max_columns', None)
df_acidentes = pd.read_csv("https://cognitrilha.com.br/datasets/demostrativo_acidentes_eco050.csv.gz", encoding= "iso-8859-1", sep=";", decimal = ",")
df_eco = pd.read_csv("https://cognitrilha.com.br/datasets/demostrativo_acidentes_eco101.csv.gz", encoding= "iso-8859-1", sep=";", decimal = ",")
df_ecosul = pd.read_csv("https://cognitrilha.com.br/datasets/demostrativo_acidentes_ecosul.csv.gz", encoding= "iso-8859-1", sep=";", decimal = ",")
```

Sobre o dataframe unificado, faça um gráfico de linhas que mostre a evolução de acidentes ano a ano, desde o início do período, onde cada linha representa um trecho da estrada, os valores são a média de total de vítimas por acidente, e cada linha é exibida em uma cor diferente.

```python
#Crie 3 DataFrames do pandas com os arquivos acima.
display(df_eco.head())
display(df_ecosul.head())
display(df_acidentes.head())
```

```python
#Crie um DataFrame com a junção dos outros 3 (use pd.concat)
df_unificado = pd.concat([df_acidentes, df_eco, df_ecosul])
df_unificado.head()
```

```python
#Normalize os dados da coluna "tipo_de_ocorrencia"
display(df_unificado["tipo_de_ocorrencia"].unique())
df_unificado["tipo_de_ocorrencia_norm"] = df_unificado["tipo_de_ocorrencia"]
df_unificado["tipo_de_ocorrencia_norm"] = df_unificado["tipo_de_ocorrencia_norm"].replace({"sem vítima": "Sem vitima", "Sem vítima": "Sem vitima", "Acidente s": "Sem vitima", "com vítima": "Com vitima", "Com vítima": "Com vitima", "Acidente c": "Com vitima", "Atropelame": "Com vitima"})
display(df_unificado["tipo_de_ocorrencia_norm"].unique())
```

```python
#Normalize os dados da coluna "sentido"
df_unificado["sentido"] = df_unificado["sentido"].replace({"Crescente": "Sul", "Decrescente": "Norte"})
df_unificado["sentido"].unique()
```

```python
#Remova as linhas com valores nulos.
df_unificado = df_unificado.dropna()
df_unificado["sentido"].unique()
```

```python
#Crie uma coluna "vítimas_totais" com a soma de todas as contagens de vítimas para cada acidente
df_unificado["vitimas_totais"] = df_unificado["levemente_feridos"] + df_unificado["moderadamente_feridos"] + df_unificado["gravemente_feridos"]
df_unificado.head()
```

```python
#Mostre um gráfico de pizza com os percentuais dos estados das vítimas (ilesos, levemente feridos, moderadamente feridos, gravemente feridos e mortos)
#para todos os acidentes do DataFrame que possuem ao menos uma vítima
df_unificado_filtrado = df_unificado[df_unificado["tipo_de_ocorrencia_norm"] == "Com vitima"]
df_grafico_pizza = (df_unificado_filtrado[["ilesos", "levemente_feridos", "moderadamente_feridos", "gravemente_feridos", "mortos"]].sum() / df_unificado_filtrado[["ilesos", "levemente_feridos", "moderadamente_feridos", "gravemente_feridos", "mortos"]].sum().sum()*100)
df_grafico_pizza = df_grafico_pizza.reset_index()
df_grafico_pizza["index"] = df_grafico_pizza["index"].replace({"ilesos": "Ilesos", "levemente_feridos": "Levemente feridos", "moderadamente_feridos": "Moderadamente feridos", "gravemente_feridos": "Gravemente feridos", "mortos": "Mortos"})
df_grafico_pizza
```

```python
fig = px.pie(df_grafico_pizza, values= 0, names="index")
fig.show()
```

```python
#Mostre a quantidade de acidentes totais de cada trecho, em um gráfico de barras empilhado,
#onde as pilhas representam o sentido (norte vs sul).
#Dica: utilize df.pivot + df.plot.bar(..., stacked=True)

df_barras_empilhadas = df_unificado.groupby("trecho")["sentido"].value_counts()
df_barras_empilhadas = df_barras_empilhadas.reset_index()
fig = px.bar(df_barras_empilhadas, x="trecho", y="count", color="sentido", barmode= "stack")
fig.show()
```

```python
#Sobre o dataframe unificado, faça um gráfico de linhas que mostre a evolução de acidentes ano a ano,
#desde o início do período, onde cada linha representa um trecho da estrada,
#os valores são a média de total de vítimas por acidente, e cada linha é exibida em uma cor diferente.
df_unificado = df_unificado.dropna(subset = "data")
df_unificado["ano"] = pd.to_datetime(df_unificado["data"], format= "%d/%m/%Y").dt.year
df_grafico_linhas = df_unificado[(df_unificado["ano"]>= 2015) & (df_unificado["ano"]<= 2024)]
df_grafico_linhas = df_grafico_linhas.groupby(["trecho", "ano"])["vitimas_totais"].mean().reset_index()
fig = px.line(df_grafico_linhas, x="ano", y= "vitimas_totais", color= "trecho")
fig.show()
```
