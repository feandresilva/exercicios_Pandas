```python
#Exiba as 10 primeiras linhas da tabela
df_matriculas.head(10)
```

```python
#Exiba os dados usando .info() e .describe()
display(df_matriculas.info())
display(df_matriculas.describe())
```

```python
#Faça gráficos de pizza para as seguintes variáveis qualitativas nominais: tp_modalidade_ensino
df_modalidades = df_matriculas["tp_modalidade_ensino"].value_counts()
df_modalidades = df_modalidades.reset_index()
df_modalidades = df_modalidades.rename(columns={"count": "valores"})
fig = px.pie(df_modalidades, values="valores", names="tp_modalidade_ensino")
fig.show()
```

```python
#Faça gráficos de pizza para as seguintes variáveis qualitativas nominais: tp_rede
df_tipo_rede = df_matriculas["tp_rede"].value_counts()
df_tipo_rede = df_tipo_rede.reset_index()
df_tipo_rede = df_tipo_rede.rename(columns={"count": "valores", "tp_rede": "tipo_rede"})
fig = px.pie(df_tipo_rede, values="valores", names="tipo_rede")
fig.show()
```

```python
#Faça gráficos de pizza para as seguintes variáveis qualitativas nominais: grau_academico
df_grau_academico = df_matriculas["grau_academico"].value_counts()
df_grau_academico = df_grau_academico.reset_index()
df_grau_academico = df_grau_academico.rename(columns={"count": "valores"})
fig = px.pie(df_grau_academico, values="valores", names="grau_academico")
fig.show()
```

```python
#Faça gráficos de pizza para as seguintes variáveis qualitativas nominais: sexo
df_sexo = df_matriculas["sexo"].value_counts()
df_sexo = df_sexo.reset_index()
df_sexo = df_sexo.rename(columns={"count": "valores"})
fig = px.pie(df_sexo, values="valores", names="sexo")
fig.show()
```

```python
#Faça um gráfico de barras empilhado (bar stacked) que mostre a evolução das inscrições por grau_academico
df_evolucao_academica = df_matriculas.groupby("nu_ano_censo")["grau_academico"].value_counts()
df_evolucao_academica = df_evolucao_academica.reset_index()
df_evolucao_academica = df_evolucao_academica.rename(columns={"count": "Valores", "nu_ano_censo": "Ano"})
fig = px.bar(df_evolucao_academica, x="Ano", y="Valores", color= "grau_academico")
fig.show()
```

```python
#Exiba a correlação entre as colunas grau_academico e sexo (utilize pd.crosstab)
df_correlacao = pd.crosstab(df_matriculas["grau_academico"], df_matriculas["sexo"])
df_correlacao
```

```python
#Exiba a correlação entre as colunas grau_academico e sexo.
#Em seguida, utilize o scipy.stats.chi2_contingency para validar a correlação entre essas variáveis.
#Considere p < 0.05 → existe evidência de associação.
#p >= 0.05 → não há evidência de associação.

df_academico_modalidade_correlacao = pd.crosstab(df_matriculas["grau_academico"], df_matriculas["sexo"])
res = chi2_contingency(df_academico_modalidade_correlacao)
res.pvalue
```

```python
#Obtenha o valor v de Cramér para medir a correlação entre as variáveis grau_academico e sexo.

#Considere 0 <= V <= 1;
#0 = independência perfeita;
#1 = associação perfeita;

#Utilize scipy.stats.contingency.association com method="cramer".

association(df_academico_modalidade_correlacao, method="cramer")
```

```python
#Exiba a tabela unificada que mostre os nomes dos cursos (utilize a lista de codigos do inep: https://repositorio.seade.gov.br/dataset/cb99a06d-7b66-4e58-bd3f-2a7e52fa8607/resource/b6abe70a-3cff-41a7-935f-23892d1c02b8/download/cine_inep.csv)

df_matriculas2 = pd.read_csv("https://cognitrilha.com.br/datasets/cine_inep.csv", encoding= "iso-8859-1", sep=";")
df_matriculas_unificado = df_matriculas.merge(df_matriculas2, on="co_cine_rotulo", how="left")
df_matriculas_unificado
```

```python
#Exiba o top 5 cursos por quantidade de novas matrículas.
df_matriculas_unificado
df_top_cursos = df_matriculas_unificado.groupby("no_cine_rotulo")["n_matricula"].sum().sort_values(ascending=False)
df_top_cursos.head(5)
```

```python
#Exiba o maior curso por quantidade de matrículas para cada grau academico
df_por_grau = df_matriculas_unificado.groupby(["grau_academico", "no_cine_rotulo"])["n_matricula"].sum().sort_values(ascending=False)
df_por_grau
```

```python
#Crie um dataframe a partir do dataset abaixo e exiba as 5 primeiras linhas
#https://cognitrilha.com.br/datasets/DTB_2024/RELATORIO_DTB_BRASIL_2024_MUNICIPIOS.xls

df_municipios = pd.read_excel("https://cognitrilha.com.br/datasets/DTB_2024/RELATORIO_DTB_BRASIL_2024_MUNICIPIOS.xls", skiprows=6)
df_municipios.head(5)
```

```python
#Unifique o dataframe de matrículas com o dataframe de municípios, com base na coluna cod_ibge.
df_municipios = df_municipios.rename(columns={"Código Município Completo": "cod_ibge"})
df_matriculas_localidade = df_matriculas_unificado.merge(df_municipios, on="cod_ibge", how="left")
df_matriculas_localidade["Nome_UF"].unique()
```

```python
#Exiba o top 5 cursos por número de matrículas do estado SP.
df_matriculas_SP = df_matriculas_localidade[df_matriculas_localidade["Nome_UF"]== "São Paulo"]
df_matriculas_SP = df_matriculas_SP.groupby("no_cine_rotulo")["n_matricula"].sum().sort_values(ascending=False)
df_matriculas_SP.head(5)
```

```python
#Exiba o top 5 cursos por número de matrículas da região Sudeste.
df_sudeste = df_matriculas_localidade[(df_matriculas_localidade["Nome_UF"] == "São Paulo") | (df_matriculas_localidade["Nome_UF"] == "Rio de Janeiro") | (df_matriculas_localidade["Nome_UF"]== "Minas Gerais") | (df_matriculas_localidade["Nome_UF"] == "Espírito Santo")]
df_sudeste["Nome_UF"].unique()
```

```python
#Crie um DataFrame a partir dos dados  no arquivo abaixo:
#https://cognitrilha.com.br/datasets/ies_2026.xls

#Exiba as 5 primeiras linhas do dataframe

df = pd.read_excel("https://cognitrilha.com.br/datasets/ies_2026.xls", skiprows=4)
df.head(5)
```

```python
#Faça um dataframe unificado (merge) através da coluna "co_ies" = "Código Mantenedora"
df = df.rename(columns={"Código IES": "co_ies"})
df_unificado2 = df.merge(df_matriculas_localidade, on="co_ies", how="inner")
df_unificado2
```

```python
#Filtre apenas as colunas:

#nu_ano_censo, tp_modalidade_ensino, tp_rede, grau_academico, sexo, n_matricula, Nome_UF, Nome_Município, Instituição(IES), Categoria, Data do Ato de Criação da IES e Situação da IES

df_unificado2_filtrado = df_unificado2[["co_ies", "nu_ano_censo", "tp_modalidade_ensino", "tp_rede", "grau_academico", "sexo", "n_matricula", "Nome_UF", "Nome_Município", "Instituição(IES)", "Categoria", "Data do Ato de Criação da IES", "Situação da IES"]]
df_unificado2_filtrado
```

```python
#Faça um df_final: Filtre apenas as linhas onde a coluna Situação da IES == "Ativa".
df_final = df_unificado2_filtrado[df_unificado2_filtrado["Situação da IES"] == "Ativa"]
df_final.head(5)
```

```python
#Agrupe por ano e IES e some o numero de matriculas, para exibir a IES que teve o maior número de matrículas de cada ano.
df_IES_mais_matriculas = df_final.groupby(["nu_ano_censo", "Instituição(IES)"])["n_matricula"].sum().sort_values(ascending=False)
df_IES_mais_matriculas = df_IES_mais_matriculas.reset_index()
df_IES_mais_matriculas
```

```python
#Agora, exiba as IES publicas com o maior número de matrículas de cada ano.
df_IES_publicas = df_final[df_final["Categoria"] == "Publica"]
df_IES_publicas_mais_matriculas = df_IES_publicas.groupby(["nu_ano_censo", "Instituição(IES)"])["n_matricula"].sum().sort_values(ascending=False)
df_IES_publicas_mais_matriculas
```
