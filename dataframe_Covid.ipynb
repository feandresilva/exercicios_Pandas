Parte 3: Operações com Datas

Extraia o ano e o mês da coluna data (use o pd.to_datetime e .dt.year e .dt.month).

Faça uma tabela agrupada de novos casos confirmados por mês/ano.

Faça um gráfico da evolução de casos ao longo do tempo (por mês/ano)

Faça um gráfico da Prevalência ao longo do tempo. (Prevalência = acumulado de casos confirmados / população estimada)

Faça um gráfico de barras que mostre a quantidade de casos confirmados agrupados por estado, em ordem do maior para o menor.

Faça um gráfico de barras que mostre a população estimada de cada estado.

Faça um gráfico em barras que mostre estado X "população estimada" e "população estimada 2019"

Faça um gráfico de pizza que mostre a proporção de casos confirmados por região (nordeste, sudeste, sul, norte, centro-oeste)

```python
#Extraia o ano e o mês da coluna data (use o pd.to_datetime e .dt.year e .dt.month).
df["ano"] = pd.to_datetime(df["date"]).dt.year
df["mês"] = pd.to_datetime(df["date"]).dt.month
df.head()
```

```python
#Faça uma tabela agrupada de novos casos confirmados por mês/ano.
df.groupby(["ano", "mês"])["new_confirmed"].sum().head()
```

```python
#Faça um gráfico da evolução de casos ao longo do tempo (por mês/ano)
df.groupby(["ano", "mês"])["new_confirmed"].sum().plot()
```

```python
#Faça um gráfico da Prevalência ao longo do tempo. (Prevalência = acumulado de casos confirmados / população estimada)
group = df.groupby(["ano", "mês"])[["new_confirmed"]].sum()
group["populacao_total"] = df.groupby("city")["estimated_population"].max().sum()
group["casos_acumulados"] = group["new_confirmed"].cumsum()
group["prevalência"] = group["casos_acumulados"] / group["populacao_total"] * 100
group["prevalência"].plot.bar()
group.tail()
```

```python
#Faça um gráfico de barras que mostre a população estimada de cada Estado
df_unico = df.drop_duplicates("city")
df_unico.groupby("state")["estimated_population"].sum().sort_values().plot.bar()
```

```python
#Faça um gráfico de barras que mostre a quantidade de casos confirmados agrupados por estado, em ordem do maior para o menor.
df.groupby("state")["new_confirmed"].sum().sort_values(ascending = False).plot.bar()
```

```python
#Faça um gráfico em barras que mostre estado X "população estimada" e "população estimada 2019"
df_unico[["estimated_population", "estimated_population_2019", "state"]].groupby("state").sum().sort_values("estimated_population", ascending = True).plot.barh()
```

```python
#Faça um gráfico de pizza que mostre a proporção de casos confirmados por região (nordeste, sudeste, sul, norte, centro-oeste)
df["regiao"] = pd.NA
df.loc[df["state"].isin(("PR", "RS", "SC")), "regiao"] = "Sul"
df.loc[df["state"].isin(("SP", "RJ", "MG", "ES")), "regiao"] = "Sudeste"
df.loc[df["state"].isin(("AL", "BA", "CE", "MA", "PB", "PE", "PI", "RN", "SE")), "regiao"] = "Nordeste"
df.loc[df["state"].isin(("AC", "AP", "AM", "PA", "RO", "RR", "TO")), "regiao"] = "Norte"
df.loc[df["state"].isin(("DF", "GO", "MT", "MS")), "regiao"] = "Centro-Oeste"
df.groupby("regiao")["new_confirmed"].sum().plot.pie()
```

```python
#Monte uma pivot_table, mostrando:
#As linhas como município
#As colunas como semana_epidemiologica
#O valor agregado como a média de casos confirmados
#https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html
tabela = pd.pivot_table(
    df, values="new_confirmed", index=["epidemiological_week"], columns= ["regiao"], aggfunc= "mean"
)
tabela
```

Parte 4: Exercícios Avançados

Crie uma nova coluna com a diferença percentual entre populacao_estimada e populacao_estimada_2019

Encontre os municípios com registros duplicados (use duplicated())

Mostre o top 10 municípios com maior número total de casos

Gere uma tabela final consolidando:
Município
Média de casos confirmados
População média
Percentual de registros repetidos

```python
#Crie uma nova coluna com a diferença percentual entre populacao_estimada e populacao_estimada_2019
df["diferenca_populacional"] = (df["estimated_population"] - df["estimated_population_2019"]) * 100 / df["estimated_population_2019"]
df[df["estimated_population"] > 1000000][["city", "estimated_population_2019", "estimated_population", "diferenca_populacional"]].drop_duplicates().sort_values("diferenca_populacional", ascending = False)
```

```python
#Encontre os municípios com registros duplicados (use duplicated())
#df[df.duplicated()]["city"]
df_2 = df[["city", "date"]]
df_2["duplicated"] = df_2.duplicated()
df_2[df_2["duplicated"] == True]["city"].unique()
```

```python
#Mostre o top 10 municípios com maior número total de casos
df.groupby("city")["new_confirmed"].sum().sort_values(ascending = False).head(10)
```

```python
#Gere uma tabela final consolidando:
#Município
#Média de casos confirmados
#População média
#Percentual de registros repetidos
resultado_1 = df.groupby("city")["new_confirmed"].mean().to_frame()
resultado_2 = df.groupby("city")["estimated_population"].mean().to_frame()
resultado_3 = ((df_2.groupby("city")["duplicated"].sum() / df_2.groupby("city")["duplicated"].count()) * 100).to_frame()
resultado_1.join(resultado_2).join(resultado_3).to_excel("resultado.xlsx")
```
