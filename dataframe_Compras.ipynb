```python
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', None)

df_item_resultado = pd.read_csv("https://cognitrilha.com.br/datasets/comprasGOV-anual-VW_DM_PNCP_ITEM_RESULTADO-2021.csv.gz")
df_compra = pd.read_csv("https://cognitrilha.com.br/datasets/comprasGOV-anual-VW_FT_PNCP_COMPRA-2021.csv.gz")
df_compra_item = pd.read_csv("https://cognitrilha.com.br/datasets/comprasGOV-anual-VW_FT_PNCP_COMPRA_ITEM-2021.csv.gz")
```

**Parte 2: Operações com Datas**

Crie as colunas ano e mês a partir da coluna data_inclusao (use o pd.to_datetime e .dt.year e .dt.month).

Faça um gráfico de barras do valor_total_homologado por mês.

```python
#Crie as colunas ano e mês a partir da coluna data_inclusao (use o pd.to_datetime e .dt.year e .dt.month)
df_compra["ano"] = pd.to_datetime(df_compra["data_inclusao"]).dt.year
df_compra["mes"] = pd.to_datetime(df_compra["data_inclusao"]).dt.month
df_compra.head()
```

```python
#Faça um gráfico de barras do valor_total_homologado por mês.
df_compra.groupby(["ano", "mes"])["valor_total_homologado"].sum().plot.bar()
```

```python
#Faça um gráfico de barras da contagem de compras por mês.
df_compra.groupby("mes")["ano"].count().plot.bar()
```

**Parte 3: Análises com Join**


Crie um dataframe unificado com todos os dados de compras e de itens. Use o cod_compra como coluna de join/merge.
(Documentação: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.join.html)

Mostre a tabela unificada com todos os dados.

Mostre uma tabela da contagem de categorias (item_categoria_nome) agrupados por estado (unidade_orgao_uf_sigla)

Mostre a quantidade única de categorias por estado (use .nunique())

Mostre um gráfico de barras com a quantidade de registros  de  cada valor da coluna 'material_ou_servico_nome' agrupado por estado (use .value_counts())

```python
df_unificado = df_compra.set_index("cod_compra").join(df_compra_item.set_index("cod_compra"), rsuffix = "_i")
df_unificado.head()
```

```python
#Mostre uma tabela da contagem de categorias (item_categoria_nome) agrupados por estado (unidade_orgao_uf_sigla)
df_unificado.groupby("unidade_orgao_uf_sigla")["item_categoria_nome"].count().sort_values(ascending = False).head()
```

```python
#Mostre a quantidade única de categorias por estado (use .nunique())
df_unificado.groupby("unidade_orgao_uf_sigla")["item_categoria_nome"].nunique().head()
```

```python
#Mostre um gráfico de barras com a quantidade de registros  de  cada valor da coluna
#'material_ou_servico_nome' agrupado por estado (use .value_counts())
plt.figure(figsize=(10, 6))
df_unificado.groupby("unidade_orgao_uf_sigla")["material_ou_servico_nome"].value_counts().plot.bar()
```

```python
#Crie um dataframe unificado das demais tabelas com o df_item_resultado. Utilize a chave id_compra_item para esse join.
df_final = df_unificado.set_index("id_compra_item").join(df_item_resultado.set_index("id_compra_item"), rsuffix= "_f")
df_final
```

Faça uma tabela unificada entre o df_item_resultado e o df_compra_item usando o parâmetro `how="inner"`


Faça uma tabela unificada entre o df_item_resultado e o df_compra_item usando o parâmetro `how="left"`


Faça uma tabela unificada entre o df_item_resultado e o df_compra_item usando o parâmetro `how="outer"`


Compare a quantidade de registros das 3 tabelas.


Compare a quantidade de valores nulos das 3 tabelas.

```python
#Faça uma tabela unificada entre o df_item_resultado e o df_compra_item usando o parâmetro how="inner"
df_inner_unificado = df_item_resultado.set_index("id_compra_item").join(df_compra_item.set_index("id_compra_item"), how = "inner", rsuffix= "_inner")
df_inner_unificado.head()
```

```python
#Faça uma tabela unificada entre o df_item_resultado e o df_compra_item usando o parâmetro how="left"
df_left_unificado = df_item_resultado.set_index("id_compra_item").join(df_compra_item.set_index("id_compra_item"), how = "left", rsuffix= "_left")
df_left_unificado.head()
```

```python
#Faça uma tabela unificada entre o df_item_resultado e o df_compra_item usando o parâmetro how="outer"
df_outer_unificado = df_item_resultado.set_index("id_compra_item").join(df_compra_item.set_index("id_compra_item"), how = "outer", rsuffix= "_outer")
df_outer_unificado.head()
```

```python
#Compare a quantidade de registros das 3 tabelas.
print(len(df_inner_unificado))
print(len(df_left_unificado))
print(len(df_outer_unificado))
```

```python
#Compare a quantidade de valores nulos das 3 tabelas
display(df_inner_unificado.isna().sum())
display(df_left_unificado.isna().sum())
display(df_outer_unificado.isna().sum())
```
