```python
#Qual o estado da região Sudeste que teve o menor valor na média da “ODS 3 do Ano 2022”, organizada por “UF”?
df_sem_nulos = df_ODS.dropna(subset=["Territorialidade"])
df_sem_nulos["UF"] = df_sem_nulos["Territorialidade"].str.split(";").str[2]
df_sem_nulos["Regiao"] = df_sem_nulos["Territorialidade"].str.split(";").str[3]
df_sudeste = df_sem_nulos[df_sem_nulos["Regiao"] == "SE"]
media_estados_sudeste = df_sudeste.groupby("UF")["2022 ODS 3"].mean().sort_values()
media_estados_sudeste
```

```python
#Qual a cidade do estado que reponde corretamente o item anterior,
#e que teve o valor da “ODS ano 2022” mais próximo da média do estado inteiro?
df_sudeste_ES = df_sudeste[df_sudeste["UF"] == "ES"]
df_sudeste_ES["Diferenca"] = (68.133846 - df_sudeste_ES["2022 ODS 3"]).abs()
df_cidade_ES = df_sudeste_ES.sort_values(by="Diferenca").head(1)
df_cidade_ES
```

```python
#Qual foi a cidade do estado do Rio de Janeiro que teve o menor crescimento da “ODS 3” entre os anos 2019 e 2022
df_rio = df_sudeste[df_sudeste["UF"] == "RJ"]
df_rio["Crescimento"] = df_rio["2022 ODS 3"] - df_rio["2019 ODS 3"]
df_rio.sort_values(by= "Crescimento")
```

```python
#Aproximadamente, qual a média da "ODS 5" entre os anos 2019 e 2022 da região Nordeste?
df_nordeste = df_sem_nulos[df_sem_nulos["Regiao"] == "NE"]
df_nordeste["Media ODS 5 2019"] = df_nordeste["2019 ODS 5"].mean()
df_nordeste["Media ODS 5 2022"] = df_nordeste["2022 ODS 5"].mean()
df_nordeste_media_final = (df_nordeste["Media ODS 5 2022"] + df_nordeste["Media ODS 5 2019"]) / 2
df_nordeste_media_final.head(1)
```

```python
#Qual foi o estado da região Nordeste que teve a maior média da coluna ODS 3 do ano 2022, organizada por “UF”?
media_estados_nordeste = df_nordeste.groupby("UF")["2022 ODS 3"].mean().sort_values(ascending=False)
media_estados_nordeste
```

```python
#Qual foi a cidade do estado que responde corretamente o item anterior
#que teve o maior crescimento da Coluna ODS 3 entre os anos 2019 e 2022?
df_RN = df_nordeste[df_nordeste["UF"]== "RN"]
df_RN["Crescimento"] = df_RN["2022 ODS 3"] - df_RN["2019 ODS 3"]
df_RN.sort_values(by="Crescimento", ascending=False)
```

```python
#Aproximadamente, qual foi a média da Pontuação Índice ODS entre os anos 2019 e 2022 da região Sudeste?
df_sudeste["Media Pontuacao Indice 2019"] = df_sudeste["Pontuação Indice ODS 2019"].mean()
df_sudeste["Media Pontuacao Indice 2022"] = df_sudeste["Pontuação Indice ODS 2022"].mean()
df_sudeste_media_indice = (df_sudeste["Media Pontuacao Indice 2019"] + df_sudeste["Media Pontuacao Indice 2022"]) / 2
df_sudeste_media_indice.head(1)
```

```python
#Qual a porcentagem de cidades da região Nordeste que estão acima da média da Pontuação Índice ODS da região Sudeste (valor da questão anterior)?
df_nordeste["Media_Pontuacao_Indice_ODS"] = (df_nordeste["Pontuação Indice ODS 2022"] + df_nordeste["Pontuação Indice ODS 2019"]) / 2
df_nordeste_filtrado = df_nordeste[df_nordeste["Media_Pontuacao_Indice_ODS"] > 51.788264]
(df_nordeste_filtrado.size / df_nordeste.size) *100
```

```python
#Qual o estado da região Norte que teve o maior valor na média do ODS 3 do ano 2019?
df_norte = df_sem_nulos[df_sem_nulos["Regiao"]== "N"]
df_norte_media_ODS_3 = df_norte.groupby("UF")["2019 ODS 3"].mean().sort_values(ascending=False)
df_norte_media_ODS_3
```

```python
#Qual cidade do estado que reponde corretamente o item anterior, que teve o valor da ODS 3 do ano 2019 mais próximo da média do estado?
df_RO = df_norte[df_norte["UF"]== "RO"]
df_RO["Diferenca"] = (57.703269 - df_norte["2019 ODS 3"]).abs()
df_RO.sort_values(by="Diferenca").head(1)
```

```python
#Qual cidade do estado do Pará que teve a melhor evolução na "Classificação ODS" entre os anos 2019 e 2022?
#(considere que a classificação ODS é um ranking)
df_Para = df_norte[df_norte["UF"]== "PA"]
df_Para["Evolucao"] = df_Para["Classificação 2022"] - df_Para["Classificação 2019"]
df_Para.sort_values(by="Evolucao")
```

```python
#Qual a porcentagem de cidades da região Norte que tem o valor da média da ODS 1
#entre os anos 2019 e 2022 acima da média desse mesmo parâmetro da região Sudeste?
df_sudeste["Media ODS 1 2019"] = df_sudeste["2019 ODS 1"].mean()
df_sudeste["Media ODS 1 2022"] = df_sudeste["2022 ODS 1"].mean()
df_sudeste_media_final_ODS1 = (df_sudeste["Media ODS 1 2019"] + df_sudeste["Media ODS 1 2022"])/2 #A médiado Sudeste é de 52.507254

df_norte["Media_ODS_1_2019_2022"] = (df_norte["2022 ODS 1"] + df_norte["2019 ODS 1"])/2
df_norte_filtrado = df_norte[df_norte["Media_ODS_1_2019_2022"]> 52.507254]
df_norte_filtrado
(df_norte_filtrado.size / df_norte.size)*100
```
